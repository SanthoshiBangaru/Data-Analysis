```python
import pandas as pd
import numpy as np
df = pd.read_csv("loan_approved.csv")
```

```python
mean = df['ApplicantIncome'].mean()
print("Mean: ", mean)

std = df['ApplicantIncome'].std()
print("Standard Deviation: ", std)
```

```python
df["ApplicantIncome"].head(5)
```

```python
Z1 = (5849 - mean)/std
print(Z1)
Z2 = (4583 - mean)/std
print(Z2)
Z3 = (3000 - mean)/std
print(Z3)
Z4 = (2583 - mean)/std
print(Z4)
Z5 = (6000 - mean)/std
print(Z5)
```

```python
max = df['ApplicantIncome'].max()
min = df['ApplicantIncome'].min()
print(max," - ", min," = ",max-min)
```

```python
sk = df['ApplicantIncome'].skew()
print(sk)

kr = df['ApplicantIncome'].kurtosis()
print(kr)

# Highly skewed and leptokurtic
```

```python
sk = df['CoapplicantIncome'].skew()
print(sk)

kr = df['CoapplicantIncome'].kurtosis()
print(kr)

# Highly skewed and leptokurtic
```

```python
sk = df['LoanAmount'].skew()
print(sk)

kr = df['LoanAmount'].kurtosis()
print(kr)

# Highly skewed and leptokurtic
```

```python
sk = df['Loan_Amount_Term'].skew()
print(sk)

kr = df['Loan_Amount_Term'].kurtosis()
print(kr)

# Highly skewed and leptokurtic
```

```python
df1 = pd.get_dummies(df["Property_Area"], dtype=int, prefix='PA', drop_first=True)
df = pd.concat([df, df1], axis=1).drop(['Property_Area'], axis=1)
```

```python
df1 = pd.get_dummies(df["Gender"], dtype=int, prefix='G', drop_first=True)
df = pd.concat([df, df1], axis=1).drop(['Gender'], axis=1)

df1 = pd.get_dummies(df["Married"], dtype=int, prefix='M', drop_first=True)
df = pd.concat([df, df1], axis=1).drop(['Married'], axis=1)

df1 = pd.get_dummies(df["Education"], dtype=int, prefix='E', drop_first=True)
df = pd.concat([df, df1], axis=1).drop(['Education'], axis=1)

df1 = pd.get_dummies(df["Self_Employed"], dtype=int, prefix='SE', drop_first=True)
df = pd.concat([df, df1], axis=1).drop(['Self_Employed'], axis=1)

df1 = pd.get_dummies(df["Loan_Status (Approved)"], dtype=int, prefix='LS', drop_first=True)
df = pd.concat([df, df1], axis=1).drop(['Loan_Status (Approved)'], axis=1)
```
